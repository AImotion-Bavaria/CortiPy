```python
import numpy as np
import pandas as pd

# Path to your NPZ file
npz_path = r"C:\Users\kreilinger\Documents\CortiPy\cortipy_runs\20260707-111218_SSVEP\data.npz"

# Load the NPZ file
data = np.load(npz_path, allow_pickle=True)

print("Keys in NPZ file:")
print(data.files)

for key in data.files:
    arr = data[key]
    print(f"\n=== {key} ===")
    print(f"Shape: {arr.shape}")
    print(f"Dtype: {arr.dtype}")

    try:
        if arr.ndim == 1:
            df = pd.DataFrame(arr, columns=[key])
        elif arr.ndim == 2:
            df = pd.DataFrame(arr)
        else:
            print(f"Cannot directly display {arr.ndim}D array as a DataFrame.")
            continue

        display(df)
    except Exception as e:
        print(f"Could not convert '{key}' to DataFrame: {e}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))
plt.plot(df.iloc[:50, 0])
plt.xlabel("Sample")
plt.ylabel(df.columns[0] if df.columns[0] is not None else "Value")
plt.title("First Column")
plt.grid(True)
plt.show()
```
